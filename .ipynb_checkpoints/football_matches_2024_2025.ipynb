## Dataset for top 5 leagues + champions league in the 2024/2025 season

This is what the table would look like, a sample table of the first 5 rows is shown below

```python
import pandas as pd

df = pd.read_csv("data/football_matches_2024_2025.csv")

df.head()
```

```python
rows, columns = df.shape

print("Number of rows:", rows)
print("Number of columns:", columns)
```

These are the columns that we will be looking at, such goals scored, points earned at home and away to be able to analyze the impact of playing at home vs away

```python
columns_of_interest = [
    "competition_name",
    "date_utc",
    "home_team",
    "away_team",
    "fulltime_home",
    "fulltime_away",
    "match_outcome",
    "goal_difference",
    "total_goals",
    "home_points",
    "away_points",
    "matchday",
    "stage"
]

print(df[columns_of_interest].dtypes)
```

Date and time of earliest match and latest match to show the timespan of the dataset

```python
df["date_utc"] = pd.to_datetime(df["date_utc"])

print("Earliest match:", df["date_utc"].min())
print("Latest match:", df["date_utc"].max())
```

```python
print(df["competition_name"].value_counts())
```

This dataset contains football match data for the 2024-2025 season for the top 5 leagues and the champions league

```python
geography = {
    "Premier League": "England",
    "La Liga": "Spain",
    "Serie A": "Italy",
    "Bundesliga": "Germany",
    "Ligue 1": "France",
    "UEFA Champions League": "Europe"
}

for competition in df["competition_name"].unique():
    print(competition, "-", geography.get(competition, "Unknown"))
```
