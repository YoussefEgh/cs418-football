## CS418 Dataset - 2020/2021 Premier League Season Stats

```python
import pandas as pd

df = pd.read_csv('data/epl_2020.csv')

rows, cols = df.shape
print(f"Number of rows: {rows}")
print(f"Number of columns: {cols}")
```

**This dataset covers the entire 2020/2021 premier league season. That is shown below by extracting the date columnns**

```python
df['Date'] = pd.to_datetime(df['Date'], dayfirst = True)
start_date = df['Date'].min().date()
end_date = df['Date'].max().date()
print(f"Season Start: {start_date}")
print(f"Season End: {end_date}")
```

**Primary Columns of Interest & Data Types:**
To answer our research question regarding home field advantage, we don't need all 100+ columns (many of which are betting odds). We care specifically about match outcomes, goals, and offensive generation metrics. 

Here are the target columns and their data types:
* `HomeTeam` / `AwayTeam` (Object/String): Names of the competing clubs.
* `FTR` (Object/String): Full Time Result (H=Home Win, D=Draw, A=Away Win).
* `FTHG` / `FTAG` (Int64): Full Time Home/Away Goals.
* `HS` / `AS` (Int64): Home/Away Shots taken.
* `HST` / `AST` (Int64): Home/Away Shots on Target.

```python
# Columns we care about
columns_of_interest = ['HomeTeam', 'AwayTeam', 'FTR', 'FTHG', 'FTAG', 'HS', 'AS', 'HST', 'AST']

# Filter the dataframe
target_df = df[columns_of_interest]
print(target_df.dtypes)

display(target_df)
```
